```python
import pandas as pd
file_path = "/Users/ruyi/Desktop/NAIP/dataset/cleaned_high_cited_papers_keywords_with_scores.csv"  # 请将 'your_file.csv' 替换为你的文件路径
data = pd.read_csv(file_path)
# 显示每列的基本信息
print("每列的基本信息：")
print(data.info())

# 显示每列的统计信息
print("\n每列的统计信息：")
print(data.describe(include="all"))

# 检查每列的非空值数量和空值比例
print("\n每列的空值分析：")
missing_info = data.isnull().sum()
total_rows = data.shape[0]
missing_percentage = (missing_info / total_rows) * 100

missing_data = pd.DataFrame({
    'Missing Values': missing_info,
    'Percentage Missing': missing_percentage
}).sort_values(by='Missing Values', ascending=False)

print(missing_data)
```

```python

# 删除任意一列中存在空值的行
data = data.dropna()
print("\n已删除任意一列为空值的行.")
print(f"剩余行数: {data.shape[0]}")

# 将处理后的数据保存到新的 CSV 文件中
output_file_path = "/Users/ruyi/Desktop/NAIP/dataset/cleaned_high_cited_papers_keywords_with_scores.csv"  # 替换为你希望保存的文件名
data.to_csv(output_file_path, index=False)
print(f"处理后的数据已保存到文件: {output_file_path}")
```
