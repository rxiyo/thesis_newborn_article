```python
import pandas as pd
import numpy as np

# 加载CSV文件
# 请确保将 'your_file.csv' 替换为您的实际文件名
df = pd.read_csv('/Users/ruyi/Desktop/NAIP/dataset/naip.csv')

# 检查 TNCSISP 列是否存在
if 'TNCSISP' not in df.columns:
    raise ValueError("CSV 文件中没有找到 'TNCSISP' 列，请检查文件内容。")

# 将 TNCSISP 列分为 10 个等分区间
df['Interval'] = pd.cut(df['TNCSISP'], bins=np.linspace(0, 1, 11), include_lowest=True)

# 计算每个区间中的最小样本数
min_samples = df['Interval'].value_counts().min()

# 从每个区间中随机抽取 min_samples 个数据
samples_to_take = min_samples + 200
sampled_df = (
    df.groupby('Interval')
    .apply(lambda group: group if len(group) <= samples_to_take else group.sample(n=samples_to_take, random_state=42))
    .reset_index(drop=True)
)

# 移除辅助列“Interval”（如不需要）
sampled_df = sampled_df.drop(columns=['Interval'])

# 保存新的数据集到文件 (可选)
sampled_df.to_csv('/Users/ruyi/Desktop/NAIP/dataset/naip_box.csv', index=False)

# 显示新生成的数据集
display(sampled_df)
```
