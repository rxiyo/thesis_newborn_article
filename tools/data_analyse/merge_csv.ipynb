```python
# %% [markdown]
# # 去重操作示例
# 
# 使用 Pandas 根据 'title' 和 'published' 列去重，并查看有多少行重复。

# %%
import pandas as pd

file_path = "/Users/ruyi/Desktop/NAIP/dataset/naip.csv"  # 替换为你的文件路径
data = pd.read_csv(file_path)

# 创建 DataFrame
df = pd.DataFrame(data)
# print("原始数据：")
# print(df)

# %%
# 标记重复行（基于 'title' 和 'published' 两列）
duplicates = df[df.duplicated(subset=["title", "published"], keep=False)]

# 去重统计
duplicate_count = len(duplicates) // 2  # 每个重复值出现两次，只需统计组数即可

print("\n重复行数：", duplicate_count)
print("\n重复行的具体内容：")
print(duplicates)

# 去重后的数据
deduplicated_data = df.drop_duplicates(subset=["title", "published"], keep="first")
deduplicated_data.to_csv("/Users/ruyi/Desktop/NAIP/dataset/deduplicated_file.csv", index=False)
# print("\n去重后的数据：")
# print(deduplicated_data)
```
