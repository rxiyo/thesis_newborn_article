```python
import matplotlib.pyplot as plt
import numpy as np

# Qwen2的数据
sizes = ['0.5B', '1.5B', '7B']
mae = [0.2249, 0.2288, 0.2182]
ndcg = [0.7410, 0.7649, 0.8460]

# 转为np.array便于反转y轴
mae = np.array(mae)
ndcg = np.array(ndcg)

fig, ax1 = plt.subplots(figsize=(7,4))

# 1. 左轴 MAE 曲线
color1 = 'tab:blue'
ax1.set_xlabel('Qwen')
ax1.set_ylabel('MAE', color=color1)
ax1.plot(sizes, mae, marker='o', color=color1)
ax1.tick_params(axis='y', labelcolor=color1)
ax1.set_ylim(0.215, 0.232)

# 标数值
for i, value in enumerate(mae):
    ax1.text(sizes[i], value+0.001, f'{value:.3f}', color=color1, ha='center', fontsize=11)

# 2. 右轴 NDCG 曲线
ax2 = ax1.twinx()
color2 = 'seagreen'
ax2.set_ylabel('NDCG', color=color2)
ax2.plot(sizes, ndcg, marker='s', color=color2)
ax2.tick_params(axis='y', labelcolor=color2)
ax2.set_ylim(0.740, 0.850)

# 标数值
for i, value in enumerate(ndcg):
    ax2.text(sizes[i], value-0.01, f'{value:.3f}', color=color2, ha='center', fontsize=11)

# 增加美化
ax1.set_xticks(sizes)
plt.tight_layout()
plt.show()
```

```python
# ... existing code ...
import matplotlib.pyplot as plt
import numpy as np

# Qwen2的数据
sizes = ['0.5B', '1.5B', '7B']
mae = [0.2249, 0.2288, 0.2182]
ndcg = [0.7410, 0.7649, 0.8460]

# 转为np.array便于反转y轴
mae = np.array(mae)
ndcg = np.array(ndcg)
fig, ax1 = plt.subplots(figsize=(7, 4))

# 1. 左轴 MAE 曲线
color1 = 'tab:blue'
ax1.set_xlabel('Qwen')
ax1.set_ylabel('MAE', color=color1)
ax1.plot(sizes, mae, marker='o', color=color1)
ax1.tick_params(axis='y', labelcolor=color1)

# 增加上下方的坐标冗余
# mae_padding = 0.005  # Y 轴上下增加的冗余
ax1.set_ylim(mae.min() - mae_padding, mae.max() + mae_padding)
ax1.grid(visible=True, which='both', linestyle='--', linewidth=0.5)

# 设置x轴使其刻度平均分布并增加左右方坐标冗余
ax1.set_xticks(range(len(sizes)))  # 使用 index 作为 ticks
ax1.set_xticklabels(sizes)
x_padding = 0.5  # X 轴左右增加的冗余
ax1.set_xlim(-x_padding, len(sizes) - 1 + x_padding)

# 标数值
for i, value in enumerate(mae):
    ax1.text(
        i+0.1, value + 0.001,  # 调高偏移量 0.003，避免与折线重叠
        f'{value:.3f}', color=color1, ha='center', fontsize=11
    )
# 2. 右轴 NDCG 曲线
ax2 = ax1.twinx()
color2 = 'seagreen'
ax2.set_ylabel('NDCG', color=color2)
ax2.plot(range(len(sizes)), ndcg, marker='s', color=color2)
ax2.tick_params(axis='y', labelcolor=color2)

# 增加上下方的坐标冗余
ndcg_padding = 0.02  # Y 轴上下增加的冗余
ax2.set_ylim(ndcg.min() - ndcg_padding, ndcg.max() + ndcg_padding)

# 标数值
for i, value in enumerate(ndcg):
    ax2.text(
        i+0.1, value - 0.01,  # 调低偏移量 0.015，避免与折线重叠
        f'{value:.3f}', color=color2, ha='center', fontsize=11
    )

# 防止网格重复
ax2.grid(visible=False)

# 增加美化
plt.tight_layout()
plt.show()
```
