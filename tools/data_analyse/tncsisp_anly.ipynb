```python
import pandas as pd
import matplotlib.pyplot as plt

# 加载数据
data = pd.read_csv('/Users/ruyi/Desktop/NAIP/dataset/naip_box.csv')

# 确保 'published' 列是日期格式
data['published'] = pd.to_datetime(data['published'])

# 提取年份
data['year'] = data['published'].dt.year

# 按年份分组并计算平均值
grouped_data = data.groupby('year')[['TNCSI', 'TNCSISP']].mean().reset_index()

# 绘制对比折线图
plt.figure(figsize=(10, 6))
plt.plot(grouped_data['year'], grouped_data['TNCSI'], marker='o', label='Average TNCSI')
plt.plot(grouped_data['year'], grouped_data['TNCSISP'], marker='o', label='Average TNCSISP')

# 添加图表标题和标签
plt.title('Average TNCSI and TNCSISP by Publication Year')
plt.xlabel('Year')
plt.ylabel('Average Values')
plt.legend()
plt.grid(True)

# 显示图表
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data from the provided table
targets = ['$\mathrm{TNCSI}_{\mathrm{SP}}$', 'TNCSI']
years = ['2020', '2021', '2022', 'All years']

# NDCG values for TNCSI and TNCSISP
ndcg_tncsi = [0.673, 0.574, 0.324, 0.469]
ndcg_tncsisp = [0.869, 0.666, 0.672, 0.836]

# Set up the figure and axis
fig, ax = plt.subplots(figsize=(8, 6))

# Bar width and positions
bar_width = 0.15
index = np.arange(len(targets))

# Define colors and hatch patterns based on the original figure
colors = ['#ccebc5', "#fbe5ae", '#b3cde3', '#decbe4']
hatches = ['o', '*', 'O', '']

# Plotting the bars
for i, year in enumerate(years[:-1]):
    values = [ndcg_tncsisp[i], ndcg_tncsi[i]]
    bars = ax.bar(index + i * bar_width, values, bar_width, label=year, color=colors[i], hatch=hatches[i], edgecolor='white')
    
    # Add text labels above the bars
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height:.3f}',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=12)

# Plotting the 'All years' as a dashed line
for i, target in enumerate(targets):
    ax.hlines(y=[ndcg_tncsisp[-1], ndcg_tncsi[-1]][i], xmin=index[i] - bar_width/2, xmax=index[i] + 2.5*bar_width, 
              colors='lightcoral', linestyles='dashed', linewidth=2, label='All years' if i == 0 else "")

# Formatting the plot
ax.set_ylabel('NDCG@20', fontsize=14)
ax.set_xticks(index + bar_width)
ax.set_xticklabels(targets, fontsize=14)
ax.set_ylim(0, 1.0) # Adjust y-axis limit based on NDCG values
ax.grid(axis='y', linestyle='-', alpha=0.7)

# Legend
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles, labels, fontsize=12, loc='upper right', frameon=False)

# Add title/caption text similar to the original figure
# plt.figtext(0.5, -0.05, "Figure 5: Impact of different prediction targets on perfor-\nmance: TNCSISP demonstrates superior performance over\nTNCSI with training data from different years.", 
#             wrap=True, horizontalalignment='center', fontsize=14)

plt.tight_layout()
plt.show()
```
